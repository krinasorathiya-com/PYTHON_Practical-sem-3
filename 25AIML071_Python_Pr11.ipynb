116. Write a program to verify broadcasting compatibility rules for mismatched
shapes
• Apply Rule 1: Insert 1 in missing dimensions.
• Apply Rule 2: Replicate along dimensions of size 1.
• Test and resolve shape mismatch errors in real use cases.

```python
A = np.ones((2, 3))
B = np.ones((3,))

try:
    result = A + B
    print("Compatible")
    print("Result shape:", result.shape)
except ValueError:
    print("Not compatible")

# Mismatched example
C = np.ones((2, 3))
D = np.ones((2,))

try:
    result = C + D
    print("Compatible")
    print("Result shape:", result.shape)
except ValueError:
    print("Not compatible: Shape mismatch")
```

118. Write a program to create a Series from a NumPy array and verify data linkage
• Convert a NumPy array into a pandas Series.
• Observe dynamic linking between original array and Series.
• Understand mutability and reference behavior.

```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30, 40])

s = pd.Series(arr, copy=False)

print("Original NumPy array:")
print(arr)

print("\nOriginal Series:")
print(s)

arr[0] = 100

print("\nAfter changing NumPy array:")
print(arr)

print("\nSeries after changing NumPy array:")
print(s)
```

**Indexing and Value Access**

120. Write a program to access Series values using position and label indexing
• Use integer-based and label-based indexing.
• Slice Series using : notation and label lists.
• Compare Series indexing with NumPy array slicing.

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40], index=['A', 'B', 'C', 'D'])

print("Series:")
print(s)

print("\nPosition indexing:")
print(s.iloc[1])

print("\nLabel indexing:")
print(s.loc['C'])

print("\nPosition slicing:")
print(s.iloc[1:3])

print("\nLabel slicing:")
print(s.loc['B':'C'])

print("\nSelected labels:")
print(s.loc[['A', 'D']])
```

**Evaluating and Counting Series Data**

124. Write a program to check for duplicate, unique, and value counts in a Series
• Use unique() to get distinct values.
• Use value_counts() to summarize frequency.
• Compare distribution of repeated and unique values.

```python
import pandas as pd

s = pd.Series([10, 20, 10, 30, 20, 10])

print("Unique values:")
print(s.unique())

print("\nValue counts:")
print(s.value_counts())

print("\nDuplicate values:")
print(s[s.duplicated()])
```

**Handling Missing Data (NaNs)**

126. Write a program to create a Series with NaN values and detect them
• Insert np.NaN in a Series manually.
• Identify missing values using isnull() and notnull().
• Practice recognizing and printing incomplete records.

```python
import pandas as pd
import numpy as np

sales = pd.Series([10, np.nan, 30, np.nan, 50])

# Unique values
print("Unique sales quantities:")
print(sales.unique())

# null values
print("\nNull values in sales data:")
print(sales.isnull())

# total number of null values
print("\nTotal number of null values:")
print(sales.isnull().sum())

# not null values
print("\nNot null values in sales data:")
print(sales.notnull().sum())
```

127. Write a program to filter valid and missing values using Boolean masks
• Use notnull() to select valid entries.
• Use isnull() to isolate missing entries.
• Understand how these tools are useful for data cleaning.

```python
import pandas as pd
import numpy as np

sales = pd.Series([10, np.nan, 30, np.nan, 50])

# Filter valid values
print("Valid sales values:")
print(sales[sales.notnull()])

# Filter missing values
print("\nMissing sales values:")
print(sales[sales.isnull()])
```
