```python
def try_float(x):
    try:
        return float(x)
    except(TypeError,ValueError):
        return x
try_float((1,2,3,4,5))
```

```python
def divide(x,y):
    try:
        result=x/y
    except  ZeroDivisionError:
        print("Cannot divide a number by zero.")
    else:
        print("Division Successfully completed : ",result)
divide(4,2)
```
