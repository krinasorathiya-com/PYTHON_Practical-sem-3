Splitting array

```python
import numpy as np
a= np.arange(10)
print('First Array')
print(a)
print('\n')
print('splitting the array in 2 equal size arrays.')
b=np.split(a,2)
print(b)
print('\n')
```

```python
a=np.arange(10)
#splitting the array indicated first at 3,second 5 remaining in 3rd
print('splitting the array indicated first at 3,second 5 and remaining in 3rd.')
c=np.split(a,[3,5,9])
print(c)
print('\n')
```

```python
a=np.arange(9)
#splitting the array indicated first at 4,second 5-7 remaining in 3rd
print('splitting the array indicated first at 3,second 5 and remaining in 3rd.')
c=np.split(a,[2,3,5])
print(c)
print('\n')
```
