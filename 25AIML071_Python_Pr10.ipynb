# Question 100

# Question 110

Distinguish between slicing (view) and copying an array.

```python
import numpy as np
a = np.array([1, 2, 3, 4, 5])
view = a[1:4]
copy = a[1:4].copy()
view[0] = 99
copy[1] = 88
print("Original:", a)
print("View:", view)
print("Copy:", copy)
```
