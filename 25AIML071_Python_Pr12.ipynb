Question 142

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Amit', 'Riya', 'Raj', 'Neha'],
    'Marks': [80, 90, 75, 90]
})

mask = df['Marks'].isin([80, 90])

print("Boolean mask:")
print(mask)

print("\nFiltered DataFrame:")
print(df[mask])
```

Question 143

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Amit', 'Riya', 'Raj', 'Neha'],
    'Price': [1.0, 1.5, 0.8, 1.1]
})

print("Price less than 1.2:")
print(df[df['Price'] < 1.2])

print("\nPrice greater than 0.8 AND less than 1.5:")
print(df[(df['Price'] > 0.8) & (df['Price'] < 1.5)])

print("\nPrice less than 0.9 OR greater than 1.4:")
print(df[(df['Price'] < 0.9) | (df['Price'] > 1.4)])
```
