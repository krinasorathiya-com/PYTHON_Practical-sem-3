```python
x=int(input("Enter a number: "))
if(x%2==0 and x>0):
    print(f"The number {x} is even and Positive.")
elif(x%2==0 and x<0):
    print(f"The number {x} is odd and Negative.")
elif(x%2!=0 and x>0):
    print(f"The number {x} is odd and Positive.")
elif(x%2!=0 and x<0):
    print(f"The number {x} is odd and Negative.")
else:
    print(f"The number {x} is zero.")
```
