```python
x=int(input("Enter a number: "))
if(x%2==0 and x>0):
    print(f"The number {x} is even and Positive.")
elif(x%2==0 and x<0):
    print(f"The number {x} is odd and Negative.")
elif(x%2!=0 and x>0):
    print(f"The number {x} is odd and Positive.")
elif(x%2!=0 and x<0):
    print(f"The number {x} is odd and Negative.")
else:
    print(f"The number {x} is zero.")
```

```python
list=[1,2,3,"Bird",4,5,"Animal"]
sum = 0
for item in list:
    if isinstance(item, (int, float)):
        sum += item

print("Sum =", sum)
```

```python
list = [10, "Apple", 20, "Banana", 30]
total = 0

for item in list:
    try:
        total += item
    except:
        continue

print("Sum =", total)
```

Separate different data type list

```python
i_list = [10, "Apple", 20, "Banana", 30.678, 40.123, "Orange", 50,[3,4]]
int_list = [i for i in i_list if type(i) == int]
float_list = [i for i in i_list if type(i) == float]
string=[i for i in i_list if type(i) == str]
print("Integer List:", int_list)
print("Float List:", float_list)
print("String List:", string)
"""separate the values given in list based on data types and create separate lists for each data type."""
```
