```python
#2.Index Error in List Access
my_list=[10,20,30,40,50]
try:
    index=int(input("Enter the index of the element you want to access(0-4):"))
    print("Value:",(my_list[index]))
except IndexError:
    print("Index out of range.")
```

```python
#3.File Reading with FileNotFoundError
filename=input("Enter the file name to read:") 
try:
    with open(filename, 'r') as file:
        content=file.read()
        print(filename)
        print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
```

```python
#4.Using Finally Block
try:
    number=int(input("Enter a number:"))
    result=10/number
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except ValueError:
    print("Invalid input! Please enter a numeric value.")
else:
    print("The result of division is:", result)
finally:
    print("Execution completed.")
```

```python
#5.Handling Multiple Errors in One Block
try:
    num1=int(input("Enter the value of num1:"))
    num2=int(input("Enter the value of num2:"))
    print("Result:",num1/num2)
except (ZeroDivisionError, ValueError) as e:
    print("Error occurred:", e)
```
