```python
#2.Index Error in List Access
my_list=[10,20,30,40,50]
try:
    index=int(input("Enter the index of the element you want to access(0-4):"))
    print("Value:",(my_list[index]))
except IndexError:
    print("Index out of range.")
```

```python
#3.File Reading with FileNotFoundError
filename=input("Enter the file name to read:") 
try:
    with open(filename, 'r') as file:
        content=file.read()
        print(filename)
        print(content)
except FileNotFoundError:
    print("File not found. Please check the file path.")
```

```python
#4.Using Finally Block
try:
    number=int(input("Enter a number:"))
    result=10/number
except ZeroDivisionError:
    print("Division by zero is not allowed.")
except ValueError:
    print("Invalid input! Please enter a numeric value.")
else:
    print("The result of division is:", result)
finally:
    print("Execution completed.")
```

```python
#5.Handling Multiple Errors in One Block
try:
    num1=int(input("Enter the value of num1:"))
    num2=int(input("Enter the value of num2:"))
    print("Result:",num1/num2)
except (ZeroDivisionError, ValueError) as e:
    print("Error occurred:", e)
```

```python
#6.Create a Custom Exception
class UnderAgeError(Exception):
    pass
age=int(input("Enter your age:"))
try:
    if age<18:
        raise UnderAgeError("You must be at least 18 years old to access this content.")
    print("Access granted.")
except UnderAgeError as e:
    print("Custom Exception:", e)
```

```python
#7.TypeError Handling in Function
def add_numbers(x, y):
    if not isinstance(x, (int, float)) or not isinstance(y, (int, float)):
        raise TypeError("Both arguments must be numbers.")
    return x + y
try:
    result=add_numbers("10", 5)
    print("Result:", result)
except TypeError as e:
    print("Error:", e)
```

```python
#8.Using else with try-except
try:
    num=int(input("Enter an even number:"))
    if num%2!=0:
        raise ValueError("The number is not even.")
except ValueError as ve:
    print("Error:", ve)
else:
    print("Thank you for entering an even number:", num)
```

```python
#9.Nested Try Blocks
try:
    a=int(input("Enter a number:"))
    try:
        b=int(input("Enter another number:"))
        print("Result:",a/b)
    except ZeroDivisionError:
        print("Cannot divide by Zero.")
except ValueError:
    print("Invalid input! Please enter numeric values.")
```
